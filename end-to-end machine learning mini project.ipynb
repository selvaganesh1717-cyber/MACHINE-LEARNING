```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("https://raw.githubusercontent.com/raveendran2862/dataset/refs/heads/main/diabetes_dataset.csv")
df.head()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace = True)
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["gender"] = le.fit_transform(df["gender"])
df['location'] = le.fit_transform(df['location'])
df['smoking_history'] = le.fit_transform(df['smoking_history'])
df.head()
```

```python
df.info()
```

```python
df.corr()
```

```python
import seaborn as sns
plt.figure(figsize=(10,10))
sns.heatmap(df.corr(),fmt = '.2f',annot=True);
```

```python
x = df.drop('diabetes',axis = 1)
y = df['diabetes']
x
```

```python
y
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.1,random_state = 9)
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
x_train
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(x_train,y_train)
```

```python
y_train_pred = lr.predict(x_test)
y_train_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
accuracy_score(y_train_pred,y_test)
```

```python
lr.fit(x_test,y_test)
```

```python
y_test_pred = lr.predict(x_train)
```

```python
accuracy_score(y_test_pred,y_train)
```

Overfitting = high train, low test

Underfitting = low train, low test

```python
train --> 95
test --> 45
```

```python
sns.heatmap(confusion_matrix(y_train_pred,y_test),annot = True,fmt = '.2f');
```

```python
sns.heatmap(confusion_matrix(y_test_pred,y_train),annot = True,fmt = '.2f');
```

```python
print(classification_report(y_train_pred,y_test))
```

```python
print(classification_report(y_test_pred,y_train))
```
