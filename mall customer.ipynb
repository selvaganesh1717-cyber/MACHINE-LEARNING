```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
import numpy as np
import warnings
warnings.filterwarnings('ignore')
```

```python
mc = pd.read_csv('https://raw.githubusercontent.com/raveendran2862/dataset/refs/heads/main/Mall_Customers.csv')
mc.head()
```

```python
mc.drop('CustomerID', axis=1, inplace=True)
mc.head()
```

```python
mc.duplicated().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder() 
mc['Gender'] = le.fit_transform(mc['Gender'])
mc.head()
```

```python
mc.describe()
```

```python
sns.barplot(mc['Gender'].value_counts());
```

```python
from sklearn.preprocessing import MinMaxScaler
mn=MinMaxScaler()
mc['Age']=mn.fit_transform(mc[['Age']])
mc.head()
```
